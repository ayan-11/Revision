```python
import pandas as pd
import re

def extract_features_flexible(text):
    features = {}
    
    def extract(pattern, text):
        match = re.search(pattern, text)
        return match.group(1).strip() if match else None

    features['heel_height'] = extract(r'Heel Height:\s*(.*)', text)
    features['footwear_type'] = extract(r'Footwear type:\s*(.*)', text)
    features['color_description'] = extract(r'Description:\s*(.*)', text)
    features['imported'] = extract(r'Imported:\s*(.*)', text)

    features['contains_slippers'] = int('slippers' in text.lower())
    features['contains_fabric'] = int('fabric' in text.lower())
    features['contains_seam'] = int('seam' in text.lower())
    features['contains_pile'] = int('pile' in text.lower())

    return pd.Series(features)
```

```python
df = data.copy()
df_extracted = df['details'].apply(extract_features_flexible)
df = pd.concat([df, df_extracted], axis=1)
```

## Training Model

```python
x=df.drop(labels='price',axis=1)
y=df['price']
```

```python
x.info()
```

```python
## Implementing Linear Regression

from sklearn.model_selection import train_test_split,GridSearchCV

x_train, x_test, y_train, y_test = train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()

model.fit(x_train,y_train)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

y_pred = model.predict(x_test)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"MAE: {mae:.2f}")
print(f"MSE: {mse:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R² Score: {r2:.2f}")
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(random_state=42)
model.fit(x_train, y_train)
y_pred = model.predict(x_test)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"MAE: {mae:.2f}")
print(f"MSE: {mse:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R² Score: {r2:.2f}")
```

```python
from xgboost import XGBRegressor
params = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 6, 9],
    'learning_rate': [0.01, 0.1, 0.2],
    'subsample': [0.8, 1],
    'colsample_bytree': [0.8, 1]
}

# Instantiate the XGBRegressor
xgb = XGBRegressor(random_state=42)

# Set up GridSearchCV
grid_search = GridSearchCV(estimator=xgb,
                           param_grid=params,
                           scoring='neg_mean_absolute_error', # Or another appropriate scoring metric
                           cv=3, # Number of cross-validation folds
                           verbose=1,
                           n_jobs=-1) # Use all available cores

# Run the grid search
grid_search.fit(x_train, y_train)

# Get the best parameters and the best model
best_params = grid_search.best_params_
best_model = grid_search.best_estimator_

print("Best Hyperparameters:", best_params)

# Make predictions with the best model
y_pred = best_model.predict(x_test)

# Evaluate the best model
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"MAE (Best Model): {mae:.2f}")
print(f"MSE (Best Model): {mse:.2f}")
print(f"RMSE (Best Model): {rmse:.2f}")
print(f"R² Score (Best Model): {r2:.2f}")
```
